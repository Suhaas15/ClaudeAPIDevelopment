```python
from anthropic import Anthropic

client = Anthropic()
model="claude-sonnet-4-6"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params={
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
#system prompt exercise
messages = []
system = """
You are a python developer who writes very concise code."""

add_user_message(messages, "Write a python function that checks a string for duplicate characters.")
answer = chat(messages, system=system)
answer
```

```python
messages = []
system = """
    You are a patient math tutor. Do not directly answer a student's questions. 
    Guide them to a solution step by step.
"""

add_user_message(messages, "How do I solve the equation 2x + 3 = 7?")
answer = chat(messages, system=system)

answer
```
